```python
from osgeo import ogr

# 読み込みたいShapefileのパス
shapefile_path = "data/empty_polygon.shp"

# ドライバーの取得（Shapefile用）
driver = ogr.GetDriverByName("ESRI Shapefile")

# Shapefileのオープン（読み取り専用）
data_source = driver.Open(shapefile_path, 0)  # 0 = 読み取り専用, 1 = 書き込み可

if data_source is None:
    print("ファイルが開けませんでした。パスを確認してください。")
    exit()

# レイヤーを取得（Shapefileは基本1レイヤー）
layer = data_source.GetLayer()
print(type(layer))
extent = layer.GetExtent()
print(layer.GetExtent)
print(extent)
```

```python
import struct

def read_shapefile_extent(shp_path):
    with open(shp_path, "rb") as f:# extent は byte 36 から double 4つ
        f.seek(36)
        data = f.read(8 * 4)
        xmin, ymin, xmax, ymax = struct.unpack("<4d", data)
    return xmin, ymin, xmax, ymax
shp = "data/empty_polygon.shp"
extent = read_shapefile_extent(shp)

print("Extent:",extent)
```

```python
import struct

def write_extent(path, xmin, ymin, xmax, ymax):
    with open(path, "r+b") as f:
        f.seek(36)
        f.write(struct.pack("<4d", xmin, ymin, xmax, ymax))


xmin, ymin, xmax, ymax = 122.0, 20.0, 154.0, 46.0

write_extent("data/empty_polygon.shp", xmin, ymin, xmax, ymax)
write_extent("data/empty_polygon.shx", xmin, ymin, xmax, ymax)
```
